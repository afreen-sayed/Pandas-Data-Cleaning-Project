```python
# Importing necessary Python libraries
import pandas as pd
```

```python
# Loading dataset into a pandas DataFrame
# Displaying first few rows to understand structure
```

```python
data = pd.read_csv('cafe.csv', encoding='latin1')
```

```python
 # Converting to integer
```

```python
data['Quantity'] = pd.to_numeric(data['Quantity'], errors='coerce')
```

```python
data['Price Per Unit'] = pd.to_numeric(data['Price Per Unit'], errors='coerce')
```

```python
data['Total Spent'] = pd.to_numeric(data['Total Spent'], errors='coerce')
```

```python
data.info()
```

```python
# Dropping rows with missing values
```

```python
data=data.dropna(subset=['Quantity'])
```

```python
data.isnull().sum()
```

```python
# Calculating price per unit for each transaction
```

```python
data['Price Per Unit']= data['Total Spent']/ data['Quantity']
```

```python
data.isnull().sum()
```

```python
data.head(30)
```

```python
# Replace missing category with 'Unknown'
```

```python
data['Item'] = data['Item'].replace(['ERROR', ' ', '',], 'UNKNOWN')
```

```python
data.columns = data.columns.str.strip()
data['Item'] = data['Item'].fillna('UNKNOWN')
```

```python
data.isnull().sum()
```

```python
data.head(30)
```

```python
# Defining default item prices and filling missing 'Price Per Unit' values based on item name
```

```python
prices={
    'Cake':3.0,
    'Smoothie':4.0,
    'Cookie':1.0,
    'Juice':3.0,
    'Sandwich':4.0,
    'Salad':5.0,
    'Tea':1.5,
    'Coffee':2.0,
}

def fill(row):
    if pd.isna(row['Price Per Unit']):
        return prices.get(row['Item'], row['Price Per Unit'])
    else:
        return row['Price Per Unit']
```

```python
data['Price Per Unit'] = data.apply(fill, axis=1)
```

```python
data.head(50)
```

```python
data.isnull().sum()
```

```python
# Calculate total amount spent per transaction
```

```python
data['Total Spent']=data['Quantity']*data['Price Per Unit']
```

```python
data.isnull().sum()
```
